```python
import graphviz

def create_election_flowchart(output_name='flowchart', file_format='pdf'):
    # Initialize Digraph
    dot = graphviz.Digraph('ELECTION_FLOWCHART', format=file_format)
    
    # Set overall graph attributes
    dot.attr(rankdir='TB', splines='ortho', nodesep='0.6', ranksep='0.7')
    
    # Global node style
    dot.attr('node', 
             fontname='Helvetica,Arial,sans-serif',
             fontsize='11',
             style='filled',
             fillcolor='#D3E4F5',  # light professional blue
             color='#4A90E2',
             penwidth='1.2')
    
    # Global edge style
    dot.attr('edge', 
             color='#4A90E2',
             penwidth='1.2',
             arrowhead='normal',
             arrowsize='0.8')
    
    #
    # Define Nodes
    #
    dot.node('start', 
             label='Start:\nFix party threshold\nand N seats',
             shape='rectangle',
             style='filled,rounded',
             fillcolor='#E6E6E6')  # slightly different color
    
    dot.node('vote', 
             label='Voters turn out\n(and vote strategically)',
             shape='rectangle')
    
    dot.node('drop', 
             label='Remove parties\nfailing threshold',
             shape='rectangle')
    
    dot.node('dHondt', 
             label="Allocate N seats\nusing D'Hondt\namong remaining parties",
             shape='rectangle')
    
    # Decision about MEC
    dot.node('mecExists', 
             label='Is there any\nMEC?',
             shape='diamond',
             fillcolor='#F7EFC6')  # pale yellow
    
    # MEC selection branch
    dot.node('mecSelect', 
             label='Select MEC:\n1) Max unanimity\n2) Max social utility\n3) Random if tied',
             shape='rectangle')
    
    dot.node('mecPolicy', 
             label="Government policy:\nMEC's unanimous policies",
             shape='rectangle')
    
    # No MEC branch
    dot.node('minority', 
             label='Largest party attempts\nminority government\n(with non-negative utility partners)',
             shape='rectangle')
    
    # Decision about feasibility
    dot.node('minorityExists', 
             label='Minority\ngovt feasible?',
             shape='diamond',
             fillcolor='#F7EFC6')
    
    dot.node('minorityPolicy', 
             label='Adopt unanimous policy\nof largest party + partners',
             shape='rectangle')
    
    # Rerun if no govt
    dot.node('rerun', 
             label='No feasible\ngovernment:\nElection fails,\n must rerun',
             shape='rectangle',
             fillcolor='#FDECEA',  # light red
             color='#E74C3C')
    
    #
    # Define Edges
    #
    dot.edge('start', 'vote')
    dot.edge('vote', 'drop')
    dot.edge('drop', 'dHondt')
    dot.edge('dHondt', 'mecExists')
    
    # MEC branch
    dot.edge('mecExists', 'mecSelect', label='Yes', fontsize='10')
    dot.edge('mecSelect', 'mecPolicy')
    
    # No MEC branch
    dot.edge('mecExists', 'minority', label='No', fontsize='10')
    dot.edge('minority', 'minorityExists')
    dot.edge('minorityExists', 'minorityPolicy', label='Yes', fontsize='10')
    dot.edge('minorityExists', 'rerun', label='No', fontsize='10')
    
    # Save and render to file
    dot.render(output_name, cleanup=True)
    print(f"Flowchart generated and saved as {output_name}.{file_format}")

if __name__ == '__main__':
    create_election_flowchart('election_flowchart', 'png')
```

```python
import graphviz

def create_two_round_flowchart(output_name='two_round_flowchart', file_format='png'):
    """
    Generates a vertical flowchart for the open, two-round,
    top-two runoff rules in a districted system.
    Saves the output as a PNG by default.
    """
    dot = graphviz.Digraph('TWO_ROUND_FLOWCHART', format=file_format)
    
    # Overall graph layout: top-to-bottom, orthogonal edges
    dot.attr(rankdir='TB', splines='ortho', nodesep='0.5', ranksep='0.7')
    
    # Global node style
    dot.attr('node',
             fontname='Helvetica,Arial,sans-serif',
             fontsize='10',
             style='filled',
             fillcolor='#D3E4F5',  # light professional blue
             color='#4A90E2',
             penwidth='1.2')
    
    # Global edge style
    dot.attr('edge',
             color='#4A90E2',
             penwidth='1.2',
             arrowhead='normal',
             arrowsize='0.8')
    
    #
    # Define Nodes (short labels to keep them relatively narrow)
    #
    dot.node('start',
             label='Start:\nFix # of seats',
             shape='rectangle',
             style='filled,rounded',
             fillcolor='#E6E6E6')
    
    dot.node('partyChoice',
             label='Candidates\nchoose party\n(max. utility)',
             shape='rectangle')
    
    dot.node('strategicExit',
             label='Phased strategic exit:\n- Polling data\n- Each candidate decides\nwhether to drop out',
             shape='rectangle')
    
    dot.node('exitCondition',
             label='3 phases or\nonly 3 remain?',
             shape='diamond',
             fillcolor='#F7EFC6')
    
    dot.node('firstRound',
             label='First-round voting:\nVoters choose\npreferred candidate\n(strategically)',
             shape='rectangle')
    
    dot.node('topTwo',
             label='Top 2 proceed\n to 2nd round',
             shape='rectangle')
    
    dot.node('secondRound',
             label='Second-round voting:\nNo strategic\nconsideration',
             shape='rectangle')
    
    dot.node('policyModel',
             label='Form govt policy:\n1) Simple majority rule\nor\n2) Party-based coalition',
             shape='rectangle')
    
    dot.node('end',
             label='End',
             shape='rectangle',
             style='filled,rounded',
             fillcolor='#E6E6E6')
    
    #
    # Define Edges
    #
    dot.edge('start', 'partyChoice')
    dot.edge('partyChoice', 'strategicExit')
    dot.edge('strategicExit', 'exitCondition', label='After each phase')
    
    # Diamond to first-round
    dot.edge('exitCondition', 'firstRound', label='Yes')
    
    # If not yet done, go back for next phase
    dot.edge('exitCondition', 'strategicExit', label='No', style='dashed')
    
    # After first round -> top two -> second round
    dot.edge('firstRound', 'topTwo')
    dot.edge('topTwo', 'secondRound')
    
    # After second round -> policy formation
    dot.edge('secondRound', 'policyModel')
    
    # End
    dot.edge('policyModel', 'end')
    
    # Render to file
    dot.render(output_name, cleanup=True)
    print(f"Two-round flowchart generated and saved as {output_name}.{file_format}")

if __name__ == '__main__':
    create_two_round_flowchart('two_round_flowchart', 'png')
```
