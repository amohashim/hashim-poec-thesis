```python
import re
import pandas as pd

def clean_data(df):
    """
    Cleans a dataframe by removing wrappers like '[...]', 'String[type]', and converting data to appropriate formats.
    
    Args:
    - df (pd.DataFrame): Input dataframe with wrapped data.

    Returns:
    - pd.DataFrame: Cleaned dataframe with converted data types.
    """
    def clean_value(value):
        if isinstance(value, str):
            # Remove list-like wrappers: "[value]"
            value = re.sub(r'^\[(.*)\]$', r'\1', value)
            # Remove type-like wrappers: "String[type]" or "Bool[value]"
            value = re.sub(r'^String\d*\["(.*)"\]$', r'\1', value)
            value = re.sub(r'^Bool\[(.*)\]$', r'\1', value)
            # Attempt to convert to numeric if possible
            try:
                return float(value) if '.' in value else int(value)
            except ValueError:
                return value
        return value

    # Apply the cleaning function to all elements in the dataframe
    return df.applymap(clean_value)
```

```python
import os
import pandas as pd

def combine_csv_files(folder_path):
    """
    Combines all CSV files in the specified folder into a single DataFrame.

    Parameters:
        folder_path (str): The path to the folder containing the CSV files.

    Returns:
        pd.DataFrame: A DataFrame containing the combined data from all CSV files.
    """
    # Initialize an empty list to store DataFrames
    data_frames = []

    # Loop through each file in the folder
    for file_name in os.listdir(folder_path):
        # Build the full file path
        file_path = os.path.join(folder_path, file_name)

        # Check if the file is a CSV
        if file_name.endswith('.csv'):
            # Read the CSV file into a DataFrame
            df = pd.read_csv(file_path)

            # Append the DataFrame to the list
            data_frames.append(df)

    # Combine all DataFrames in the list
    combined_df = pd.concat(data_frames, ignore_index=True)

    return combined_df
```

```python
df_raw = combine_csv_files("/Users/alihashim/Desktop/Online_Academic_Submissions/poec_thesis/csvs")
```

```python
# Clean the dataset
cleaned_data = clean_data(df_raw)
cleaned_data.loc[:,"L"] = cleaned_data.loc[:,"L"].str.strip('"')
cleaned_data.loc[:,"M"] = cleaned_data.loc[:,"M"].str.strip('"')
cleaned_data.loc[:,"N"] = cleaned_data.loc[:,"N"].str.strip('"')

cleaned_data.to_csv("cleaned_data.csv")
# cleaned_data.to_csv("cleaned_L1_L2_L2_L1_1_to_2000.csv")
```
