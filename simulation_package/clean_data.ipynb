```python
import re
import pandas as pd

def clean_data(df):
    """
    Cleans a dataframe by removing wrappers like '[...]', 'String[type]', and converting data to appropriate formats.
    
    Args:
    - df (pd.DataFrame): Input dataframe with wrapped data.

    Returns:
    - pd.DataFrame: Cleaned dataframe with converted data types.
    """
    def clean_value(value):
        if isinstance(value, str):
            # Remove list-like wrappers: "[value]"
            value = re.sub(r'^\[(.*)\]$', r'\1', value)
            # Remove type-like wrappers: "String[type]" or "Bool[value]"
            value = re.sub(r'^String\d*\["(.*)"\]$', r'\1', value)
            value = re.sub(r'^Bool\[(.*)\]$', r'\1', value)
            # Attempt to convert to numeric if possible
            try:
                return float(value) if '.' in value else int(value)
            except ValueError:
                return value
        return value

    # Apply the cleaning function to all elements in the dataframe
    return df.applymap(clean_value)
```

```python
import os
import glob
import re
import pandas as pd

def combine_and_clean_csvs(input_dir, output_csv):
    """
    Go through each CSV in `input_dir`, clean them using regex, and write out a single combined CSV to `output_csv`.
    Specifically removes:
      - StringX["..."] tokens (e.g. String3["L2"])
      - bracketed tokens [anything]
      - Bool[...] tokens (Bool[0], Bool[1], etc.)
    """
    all_dfs = []

    # 1. Find all CSVs in the directory
    csv_files = glob.glob(os.path.join(input_dir, "*.csv"))

    for csv_file in csv_files:
        # 2. Read each CSV. 
        #    Example below: skip row #1 if it's a junk row or repeated header.
        #    If that isn't needed, remove "skiprows".
        df = pd.read_csv(csv_file, skiprows=[1])

        # -----------------------------------------------------
        # 3. Regex-based cleaning of the entire DataFrame
        # -----------------------------------------------------

        # We'll define several patterns:

        # a) Remove 'StringX["..."]' patterns, e.g. String3["L2"]
        pattern_string = r"String\d+\[\"[A-Za-z0-9_\.]+\"\]"

        # b) Remove bracketed tokens of any kind (like [0.0], [2], [0.0some], etc.)
        #    We use a simple non-greedy match from `[` to `]`.
        pattern_brackets = r"\[.*?\]" 

        # c) Remove Bool[...] tokens, e.g. Bool[0], Bool[1].
        #    Specifically matches "Bool[" plus digits or something, then a closing bracket.
        pattern_bool = r"Bool\[\d+\]"

        # Combine them into a single list of patterns to remove:
        patterns_to_remove = [pattern_string, pattern_brackets, pattern_bool]

        # --- A) Clean the data cells themselves ---
        df.replace(
            to_replace=patterns_to_remove,
            value="",
            regex=True,
            inplace=True
        )

        # --- B) Clean the column names ---
        new_cols = []
        for col in df.columns:
            col_clean = col
            for pat in patterns_to_remove:
                col_clean = re.sub(pat, "", col_clean)
            col_clean = col_clean.strip()  # remove leading/trailing spaces
            new_cols.append(col_clean)
        df.columns = new_cols

        # 4. Collect the cleaned df
        all_dfs.append(df)

    # 5. Concatenate all DataFrames
    if all_dfs:
        combined_df = pd.concat(all_dfs, ignore_index=True)
    else:
        combined_df = pd.DataFrame()

    # 6. Output final CSV
    combined_df.to_csv(output_csv, index=False)
```

```python
# Define cleaning functions
def parse_bracketed_values(value):
    """Remove brackets from values and convert to appropriate types."""
    if isinstance(value, str) and value.startswith('[') and value.endswith(']'):
        content = value.strip('[]')
        try:
            # Convert to numeric if possible
            if '.' in content:
                return float(content)
            else:
                return int(content)
        except ValueError:
            return content  # Return as string if not numeric
    return value

def parse_bool_values(value):
    """Convert Bool[x] to boolean values."""
    if isinstance(value, str) and value.startswith('Bool[') and value.endswith(']'):
        return value.strip('Bool[]') == '1'
    return value

def parse_string_types(value):
    """Extract meaningful content from String3 and String7 types."""
    if isinstance(value, str) and ('String3[' in value or 'String7[' in value):
        return value.split('[')[1].strip('"]')
    return value

df = pd.read_csv("/Users/alihashim/Desktop/Online_Academic_Submissions/poec_thesis/simulation_package/test_fix.csv")
# Apply the cleaning functions to the dataset
cleaned_data = df.copy()
for col in cleaned_data.columns:
    cleaned_data[col] = (
        cleaned_data[col]
        .apply(parse_bracketed_values)
        .apply(parse_bool_values)
        .apply(parse_string_types)
    )
```
