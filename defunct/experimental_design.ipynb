```python
import math

def fractional_design_runs(a, M):
    """
    Compute the number of treatment combinations for a fractional factorial design 
    on a three-level experiment using the formula:
    
        N(a, M) = 1 + sum_{r=1}^{M} binom(a, r)*2^r
    
    where a is the number of three-level factors and M is the maximum order 
    of interactions desired.
    
    Parameters:
    -----------
    a : int
        Number of three-level factors.
    M : int
        Maximum order of interactions to be estimated (e.g. M=3 means
        main effects, 2-way, and 3-way interactions).
        
    Returns:
    --------
    int
        The total number of unique treatment combinations.
    """
    total = 1
    for r in range(1, M+1):
        total += math.comb(a, r) * (2**r)
    return total

# Example usage:
if __name__ == '__main__':
    # For M = 3, compute design runs for a=3,...,7 active factors.
    for a in range(3, 8):
        runs = fractional_design_runs(a, 4)
        print(f"For a={a} active three-level factors and M=3, N(a,3) = {runs}")

    # Optionally, to compute the overall total for our split design above:
    # Frequencies: a=3:1, a=4:4, a=5:6, a=6:4, a=7:1.
    freqs = {3: 1, 4: 4, 5: 6, 6: 4, 7: 1}
    total_treatments = sum(freq * fractional_design_runs(a, 4) for a, freq in freqs.items())
    replications = 100
    total_simulation_runs = total_treatments * replications

    print("\nOverall treatment combinations for M:", total_treatments)
    print("Total simulation runs with 100 replications each:", total_simulation_runs)
```

```python
from math import comb

def sum_m_choose_n(m, n_start, n_end):
    """
    Calculate the sum of m choose n for a range of n values.
    
    Parameters:
    - m (int): The total number of items to choose from.
    - n_start (int): The starting value of n.
    - n_end (int): The ending value of n (inclusive).
    
    Returns:
    - int: The sum of m choose n over the range of n values.
    """
    if n_start < 0 or n_end > m or n_start > n_end:
        raise ValueError("Invalid range of n values. Ensure 0 <= n_start <= n_end <= m.")
    
    return sum(comb(m, n) for n in range(n_start, n_end + 1))

# Example usage
P = 7
n_start = 1
n_end = math.ceil(0.5 * P)
result = sum_m_choose_n(P, n_start, n_end)
print(f"The sum of m choose n from n={n_start} to n={n_end} is: {result}")
```
